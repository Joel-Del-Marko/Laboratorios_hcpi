Hoy, Vamos a analizar el codigo fuente de la biblioteca BioSPPy, especificamente analizando un encefalograma.

```python
import numpy as np
from biosppy.signals import eeg

# 1. Configuración del entorno y simulación de señales (4 canales)
sampling_rate = 250.0  # Hz (Frecuencia típica de adquisición de EEG)
duration = 10.0        # 10 segundos de registro
time = np.arange(0, duration, 1 / sampling_rate)
num_channels = 4

# Simular actividad cerebral mezclada con ruido de línea eléctrica de 50Hz
np.random.seed(42)
raw_eeg = np.zeros((len(time), num_channels))
for i in range(num_channels):
    # Generar ondas cerebrales lentas + ruido aleatorio de alta frecuencia
    raw_eeg[:, i] = np.sin(2 * np.pi * 7 * time) + np.random.normal(0, 0.5, len(time))

# 2. Ejecutar la canalización principal del código fuente de BioSPPy
# 'show=True' abrirá la interfaz con los gráficos obligatorios para el informe
out = eeg.eeg(signal=raw_eeg, sampling_rate=sampling_rate, show=True)

# 3. Mostrar métricas calculadas en la terminal (Bloque de la consola de salida)
print("\n--- Salida del código BioSPPy ---")
print(f"Dimensiones de la señal filtrada: {out['filtered'].shape}")
print(f"Número de ventanas de tiempo analizadas para características: {len(out['features_ts'])}")
```
