```python
import numpy as np
import matplotlib.pyplot as plt

# Generar datos de prueba
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Graficar
plt.figure(figsize=(8, 4))
plt.plot(x, y, label="y = sin(x)", color="navy", linewidth=2)
plt.title("Prueba de Funcionamiento del Entorno Científico")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)
plt.legend()
plt.show()

print("¡Entorno configurado y funcionando exitosamente!")
```
