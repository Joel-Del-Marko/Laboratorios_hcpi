```python
import control as ct
import matplotlib.pyplot as plt

R = 1000       # Resistencia [Ohm]
L = 0.1        # Inductancia [H]
C = 0.0001     # Capacitancia [F]

G = ct.tf([1],[L*C, R*C, 1])

print("Funcion de transferencia:")
print(G)

t1, y1 = ct.step_response(G)

plt.figure()
plt.plot(t1, y1)
plt.xlabel("Tiempo [s]")
plt.ylabel("Voltaje del capacitor [V]")
plt.title("Gráfica voltaje en el capacitor mediante la Funcion de transferencia")
plt.grid()


A = [[-R/L, -1/L],[1/C, 0]]

B = [[1/L],[0]]

C = [[0, 1]]

D = [[0]]

G_ss = ct.ss(A, B, C, D)

t2, y2 = ct.step_response(G_ss)

plt.figure()
plt.plot(t2, y2)
plt.xlabel("Tiempo [s]")
plt.ylabel("Voltaje del capacitor [V]")
plt.title("Gráfica voltaje en el capacitor mediante variables de estado")
plt.grid()


plt.figure()

plt.plot(t1, y1, label="Funcion de transferencia")
plt.plot(t2, y2, "--", label="Variables de estado")

plt.xlabel("Tiempo [s]")
plt.ylabel("Voltaje del capacitor [V]")
plt.title("Comparacion de respuestas del circuito RLC")
plt.legend()
plt.grid()

plt.show()
```
